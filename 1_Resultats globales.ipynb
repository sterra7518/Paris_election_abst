# 1 - Résultats globales des deux éléctions à Paris 

Nou allons analyser les diiférentes caractéristiques des éléctions européeenes et legislatives  à Paris 

#### Européennes 2024

```python
eur24 = pd.read_csv("res_europeennes_24.csv",sep=";",encoding="ISO-8859-1")
```

```python
#Nombre totale des abscent aux votes 
abst_eur = np.sum(eur24['nb_abscent'])

#Nombre totales d'inscrit 
insc_eur = np.sum(eur24['nb_inscr']) 


#taux de personnes abscent 
tx_abst_eur = round(abst_eur/insc_eur,3)

res_abst_eur = [insc_eur,abst_eur,tx_abst_eur]
res_abst_eur
```

```python
#Diagramme circulaire
plt.figure(figsize=(10,6))

plt.pie(partie_eur,labels = ["Extrême Gauche","Gauche","Centre","Droite","Extrême Droite","Divers"], autopct = '%.2f%%')
plt.title('Nombre de vote par tendance politiques Européennes 24')
plt.savefig("eur24_res_parties.png")

plt.show()
```

# 2 - Comparaison des abstentions 

Le but est de comparer les abstentions entre les deux élections 

```python
#On va prendre que les abstentions 

abst_eur =  pd.DataFrame([eur24["ï»¿id_bv"],eur24["nb_inscr"],eur24["nb_abscent"]]).T

abst_eur = abst_eur.rename(columns = {'nb_inscr':'nb_inscr_eur','nb_abscent':'nb_abs_eur'})


abst_eur 
```

```python
abst_leg = pd.DataFrame([leg24["ï»¿id_bv"],leg24["nb_inscr"],leg24["nb_abscent"]]).T

abst_leg = abst_leg.rename(columns = {'nb_inscr':'nb_inscr_leg','nb_abscent':'nb_abs_leg'})


abst_leg
```

#### écart d'abstentions

On a comparer les abstention entre les deux élections 

```python
abst_eur = abst_eur.set_index('ï»¿id_bv',drop='True')

abst_leg = abst_leg.set_index('ï»¿id_bv',drop='True')


abst_leg

```

```python
abst = pd.merge(abst_eur,abst_leg,on='ï»¿id_bv')

abst
```

```python
abst["nb_part_eur"] = abst["nb_inscr_eur"]-abst["nb_abs_eur"]

abst["nb_part_leg"] = abst["nb_inscr_leg"]-abst["nb_abs_leg"]

abst["écart_part"] = abst["nb_part_leg"]-abst["nb_part_eur"]

abst["rapport_part"] = np.round(abst["nb_part_leg"]/abst["nb_part_eur"],2)

abst
```

```python
abst.to_csv('abstention.csv', sep=';', encoding = 'ISO-8859-1',index=False)  
```

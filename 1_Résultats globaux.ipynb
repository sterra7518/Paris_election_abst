```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# 1 - Européennes 2024

```python
# Ouverture des fichiers 

eur24 = pd.read_csv("res_europeennes_24.csv",sep=";",encoding="ISO-8859-1")


eur24 = eur24.rename(columns = {"ï»¿id_bv":"id_bv"})

eur24 = eur24.set_index('id_bv',drop='True')


eur24
```

```python
eur24.columns
```

```python
#Nombre totale des abscent aux votes 
abst_eur = np.sum(eur24['nb_abscent'])

#Nombre totales d'inscrit 
insc_eur = np.sum(eur24['nb_inscr']) 


#taux de personnes abscent 
tx_abst_eur = round(abst_eur/insc_eur,3)

res_abst_eur = [insc_eur,abst_eur,tx_abst_eur]
res_abst_eur
```

```python
# Resultats des parties 

partie_eur = [np.sum(eur24['NB_ExtGauche']),
              np.sum(eur24['NB_Gauche']),
              np.sum(eur24['NB_Centre']),
              np.sum(eur24['NB_Droite']),
              np.sum(eur24['NB_ExtDroite']),
              np.sum(eur24['NB_Divers'])]

partie_eur
```

```python
#Diagramme circulaire
plt.figure(figsize=(10,6))

plt.pie(partie_eur,labels = ["Extrême Gauche","Gauche","Centre","Droite","Extrême Droite","Divers"], autopct = '%.2f%%')
plt.title('Nombre de vote par tendance politiques Européennes 24')
plt.savefig("eur24_res_parties.png")

plt.show()
```

# 4 - analyse des abstention 

On va analyser les distribution des taux d'abstention par bureau de vote avec les boites à moustache

```python
# distribution 

leg24["nb_abscent"] = leg24["nb_inscr"] - leg24["nb_votant"]


abst_leg24 = pd.DataFrame(leg24["nb_abscent"])

abst_leg24
```

```python
# distribution 

plt.subplot(1, 2, 1)
sns.boxplot(eur24["nb_abscent"])
plt.ylim(100,1200)
plt.title("Européennes 2024")


plt.subplot(1, 2, 2)
sns.boxplot(abst_leg24,color="green")
plt.ylim(100,1200)
plt.title("Législatives 2024")


plt.show()
```

On remarque une forte baisse des personnes qui ne sont pas allés votés et d'parès les comparaison on a vu avec les nombres voix , on a une forte mobilisation des résultats 

#### Ecart entre les bureaux de vote

```python
abst_leg24

abst_leg24 = abst_leg24.rename(columns = {'nb_abscent':'abs_leg'})

abst_leg24                
```

```python
abst_eur24 = pd.DataFrame(eur24["nb_abscent"])

abst_eur24 = abst_eur24.rename(columns = {"nb_abscent":"abs_eur"})

abst_eur24
```

```python
#Fusion des deux tables 

abst = abst_eur24.merge(abst_leg24,on="id_bv")

abst
```

```python
abst["écart"] = abst["abs_eur"] - abst["abs_leg"]

abst
```

```python
# distribution des écart d'abstention 

sns.boxplot(abst["écart"],color="green")
plt.ylim(50,500)
plt.title("distribution des écarts des abscents entre 2 élections")

plt.show()
```
